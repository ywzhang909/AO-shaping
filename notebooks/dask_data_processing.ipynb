```python
import pandas as pd
import numpy as np
import warnings
import tqdm

from PIL import Image

import swanlab
import json
```

```python
df = pd.read_pickle(r"D:\workspace\AO-shaping\data\wf-less\20251112_194423\a4506915-596f-4c55-92cd-53dd3af1e868.pkl", compression='zip')

with open(r"D:\workspace\AO-shaping\data\wf-less\20251112_194423\a4506915-596f-4c55-92cd-53dd3af1e868.json") as f:
    config = json.load(f)
    config['dump_path'] = r"D:\workspace\AO-shaping\data\wf-less\20251112_194423"
    config['experiment_date'] = "20251112"

df.head(2)
```

```python
def img_to_swan_img(img):
    return swanlab.Image(img)

def v_to_swan_bar(v):
    bar  = swanlab.echarts.Bar()
    bar.add_xaxis(list(range(len(v))))
    bar.add_yaxis("v", v.tolist())
    return bar


df['v'] = df['_v'].apply(v_to_swan_bar)
df['img'] = df['_img'].apply(img_to_swan_img)
df['diff'] = df['_diff']

df.head(2)
```

```python
swanlab.init(
    project="AO-shaping",
    experiment_name="a4506915-596f-4c55-92cd-53dd3af1e868",
    config=config,
    description="Axis Spot PIB Test"
)
```

```python
for i, row in tqdm.tqdm(df.iterrows()):
    row['v'] = v_to_swan_bar(row['_v'])
    log = {k:v for k,v in row.to_dict().items() if not k.startswith('_')}
    swanlab.log(log)
```

```python
swanlab.finish()
```
